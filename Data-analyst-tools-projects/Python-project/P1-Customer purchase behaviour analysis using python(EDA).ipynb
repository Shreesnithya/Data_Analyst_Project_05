```python
#Step 3: Customer segmentation(Groups-High value,Regular and One time buyers)
#First, build the customer summary table
customer_summary=data.groupby("CustomerName").agg(
    TotalSpend=("Spend", "sum"),
    OrderCount=("OrderID", "count"),
    AverageOrderValue=("Spend", "mean")
).reset_index()
customer_summary.head()                                                  
```

```python
#Define segmentation rules
#Rank customers by spend and then bucket into segment using percentile
customer_summary["SpendRank"]=customer_summary["TotalSpend"].rank(pct=True)

def segment_by_spend(pct):
    if pct>=0.90:
        return "High value"
    elif pct>=0.40:
        return "Regular"
    else:
        return "Onetime buyer"

customer_summary["Segment"]=customer_summary["SpendRank"].apply(segment_by_spend)
customer_summary.head()    
```

```python
#Check segment size and contribution
segment_summary=customer_summary.groupby("Segment").agg(
    NumCustomers=("CustomerName", "count"),
    TotalRevenue=("TotalSpend", "sum")
)

segment_summary["RevenueShare%"]=(segment_summary["TotalRevenue"]/customer_summary["TotalSpend"].sum()*100).round(2)
segment_summary
```

```python
#Visualize the segments
segment_counts=customer_summary["Segment"].value_counts()
plt.figure(figsize=(6,5))
sns. barplot(x=segment_counts.index,y=segment_counts.values)
plt.title("Customer Segment Distribution")
plt.xlabel("Segment")
plt.ylabel("Number of customers")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=customer_summary["Segment"],y=customer_summary["TotalSpend"],estimator=sum,errorbar=None)
plt.title("Total revenue by segment")
plt.ylabel("Total Revenue")
plt.show()
```

```python
data=data.merge(customer_summary[["CustomerName","Segment"]],on="CustomerName",how="left")
data
```

### 
1. Revenue Concentration
The project brief hypothesized that the top 10% of customers would contribute 
~60% of total revenue. In this dataset, the High Value segment (21 customers, 
~10.5% of total customers) actually contributed only **24.84%** of revenue. 
Instead, the Regular segment (100 customers, 50%) generated the majority of 
revenue (**54.85%**). This suggests spend in this dataset is fairly evenly 
distributed, rather than concentrated among a small "whale" group — likely 
because the data was randomly generated rather than reflecting the natural 
skew seen in real purchasing behavior.

### 2. Seasonality
When aggregated across all years combined, November appeared to have the 
highest total spend. However, breaking the data down year-by-year reveals 
no consistent pattern: the highest-spending month was March in 2023, 
November in 2024, and July in 2025. Since the peak month changes every 
year with no repetition, this indicates the data does not show genuine 
seasonality — the earlier "November spike" was driven by a single year's 
data rather than a recurring seasonal trend. This is consistent with the 
dataset being randomly generated rather than reflecting real seasonal 
purchasing behavior.
